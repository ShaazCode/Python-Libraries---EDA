```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[[
        "1st",
        "2nd",
        "3rd+"
    ]]
)

df[["Owner"]] = encoder.fit_transform(df[["Owner"]])

print(df.head())
```
