```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus = ["I love machine learning", "Machine learning is amazing"]
```

```python
# Inatalize countervectorize with binary ture
```

```python
vectorizer = CountVectorizer(binary=True)
```

```python
# Fit and transfoer the text data
binary_bow = vectorizer.fit_transform(corpus)
```

```python
print(binary_bow.toarray())
```
