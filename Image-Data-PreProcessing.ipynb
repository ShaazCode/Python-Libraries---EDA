```python
# Import OpenCV library for image processing
import cv2

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib to display images
import matplotlib.pyplot as plt

# Import OS library to work with folders
import os
```

```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Store the dataset folder path

dataset_path="/content/drive/MyDrive/PetImages"
```

```python
# Read one image from the dataset

#imread is used to read a image from your computer and load into python

img=cv2.imread(dataset_path+"/Cat/0.jpg")

# Check whether image is loaded successfully

if img is None:
    print("Image Not Found")
else:
    print("Image Loaded Successfully")
```

```python
# Convert image from BGR to RGB
# OpenCV reads images in BGR format
# Matplotlib displays images in RGB format

# cvtColor() → Convert Color. It changes one color format to another.
#img → The original image
# cv2.COLOR_BGR2RGB → Convert the image from BGR format to RGB format.

rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

# Display image

# imshow() → Image Show.
plt.imshow(rgb)

plt.title("Original Image")

plt.axis("off")

plt.show()
```

```python
# Print complete image shape

# print(shape) so here we will get the tuple
print("Shape :",img.shape)

# Print image height

print("Height :",img.shape[0])

# Print image width

print("Width :",img.shape[1])

# Print number of channels

print("Channels :",img.shape[2])

# Print datatype

print("Datatype :",img.dtype)
```

```python
# Resize image into 224x224

# resize() → Used to change the size of an image.
resize=cv2.resize(img,(224,224))

# Convert to RGB for display

resize_rgb=cv2.cvtColor(resize,cv2.COLOR_BGR2RGB)

plt.imshow(resize_rgb)

plt.title("Resized Image")

plt.axis("off")

plt.show()
```

```python
# Convert color image into grayscale

gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

plt.imshow(gray,cmap="gray")

plt.title("Gray Image")

plt.axis("off")

plt.show()
```

```python
# Crop image
# Format:
# image[row_start:row_end,column_start:column_end]

crop=img[100:350,100:350]

crop_rgb=cv2.cvtColor(crop,cv2.COLOR_BGR2RGB)

plt.imshow(crop_rgb)

plt.title("Cropped Image")

plt.axis("off")

plt.show()
```

```python
# Flip image horizontally

flip=cv2.flip(img,1)

flip_rgb=cv2.cvtColor(flip,cv2.COLOR_BGR2RGB)

plt.imshow(flip_rgb)

plt.title("Flipped Image")

plt.axis("off")

plt.show()
```

```python
# Rotate image by 90 degrees clockwise

rotate=cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)

rotate_rgb=cv2.cvtColor(rotate,cv2.COLOR_BGR2RGB)

plt.imshow(rotate_rgb)

plt.title("Rotated Image")

plt.axis("off")

plt.show()
```

```python
# Blur image
# Removes small noise

blur=cv2.GaussianBlur(img,(5,5),0)

blur_rgb=cv2.cvtColor(blur,cv2.COLOR_BGR2RGB)

plt.imshow(blur_rgb)

plt.title("Gaussian Blur")

plt.axis("off")

plt.show()
```

```python
# Median blur
# Good for salt and pepper noise

median=cv2.medianBlur(img,5)

median_rgb=cv2.cvtColor(median,cv2.COLOR_BGR2RGB)

plt.imshow(median_rgb)

plt.title("Median Blur")

plt.axis("off")

plt.show()
```

```python
# Remove noise but preserve edges

bilateral=cv2.bilateralFilter(img,9,75,75)

bilateral_rgb=cv2.cvtColor(bilateral,cv2.COLOR_BGR2RGB)

plt.imshow(bilateral_rgb)

plt.title("Bilateral Filter")

plt.axis("off")

plt.show()
```

```python
# Convert pixel values from 0-255 to 0-1

normalize=img/255.0

print(normalize)
```

```python
# Convert grayscale image into binary image

gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

_,binary=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)

plt.imshow(binary,cmap="gray")

plt.title("Binary Image")

plt.axis("off")

plt.show()
```

```python
# Detect edges

gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

edge=cv2.Canny(gray,100,200)

plt.imshow(edge,cmap="gray")

plt.title("Edge Detection")

plt.axis("off")

plt.show()
```
