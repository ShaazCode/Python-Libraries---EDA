```python
import cv2
# Import OpenCV library for image processing

import matplotlib.pyplot as plt
# Import Matplotlib to display images

import numpy as np
# Import NumPy for numerical operations
```

Read Image

```python
img = cv2.imread("0.jpg")
# Read the image named 0.jpg

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Convert BGR image into RGB because Matplotlib displays RGB images

plt.imshow(rgb)
# Display the image

plt.title("Original Image")
# Give title to image

plt.axis("off")
# Hide x and y axis

plt.show()
# Display the output
```

**Check Image Properties**

```python
print("Image Shape :", img.shape)
# Print Height, Width and Channels

print("Image Size :", img.size)
# Print total number of pixel values

print("Image Data Type :", img.dtype)
# Print data type of image
# stands for an 8-bit unsigned integer
```

🌍 Real-World Example 1: Portrait Mode

"By blurring the background, we focus on the important object."

```python
blur = cv2.GaussianBlur(img,(5,5),0)
# Apply Gaussian Blur
# (5,5) = Kernel Size
# 0 = Sigma value


# (5,5) = Kernel Size
# Kernel size is a small window that moves over the image.
# It tells OpenCV how many neighboring pixels to look at while blurring.
# Here, (5,5) means a window of 5 pixels wide and 5 pixels high.
# Larger kernel = More blur
# Smaller kernel = Less blur

# 0 = Sigma Value
# Sigma controls how much blur is applied.
# When we write 0, OpenCV automatically chooses the best sigma value based on the kernel size.

blur_rgb = cv2.cvtColor(blur,cv2.COLOR_BGR2RGB)

plt.imshow(blur_rgb)

plt.title("Gaussian Blur")

plt.axis("off")

plt.show()
```

Canny Edge Detection ⭐ (Most Popular)

Real-World Example
Imagine drawing only the clean and important outline of an object.

Canny removes weak edges and keeps strong ones.
Why do we use Canny?
High accuracy.
Less noise.
Produces clean edges.

Widely used in computer vision.


```python
gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
# Convert image into grayscale


edge = cv2.Canny(gray,100,200)
# Detect edges
# 100 = Lower Threshold
# 200 = Upper Threshold
# Canny() → Function used to detect the edges (boundaries) of objects in an image.
# If an edge is weaker than 100, OpenCV usually ignores it because it may be noise
# If an edge is stronger than 200, OpenCV definitely keeps it as an edge.

plt.imshow(edge,cmap="gray")

plt.title("Edge Detection")

plt.axis("off")

plt.show()
```

Image Normalization

```python
normalize = img/255.0
# Normalize pixel values between 0 and 1

print("Minimum Pixel :",normalize.min())
# Print minimum pixel value

print("Maximum Pixel :",normalize.max())
# Print maximum pixel value

normalize_rgb = cv2.cvtColor(normalize.astype("float32"),cv2.COLOR_BGR2RGB)
# Convert normalized image into RGB

plt.imshow(normalize_rgb)

plt.title("Normalized Image")

plt.axis("off")

plt.show()
```

"Computers do exactly the same thing using RGB values."

: → Select all rows.
: → Select all columns.
0 → Select Channel 0, which is the Red channel (because the image has already been converted to RGB).

1 → Green channel.

2 → Blue channel.



```python
rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
# Convert image to RGB

r = rgb[:,:,0]
# Extract Red channel

g = rgb[:,:,1]
# Extract Green channel

b = rgb[:,:,2]
# Extract Blue channel


# Stands for Color Map. It decides which color will be used to display the pixel values.
plt.imshow(r,cmap="Reds")
plt.title("Red Channel")
plt.axis("off")
plt.show()

plt.imshow(g,cmap="Greens")
plt.title("Green Channel")
plt.axis("off")
plt.show()

plt.imshow(b,cmap="Blues")
plt.title("Blue Channel")
plt.axis("off")
plt.show()
```
