```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

important_words = ["not", "no", "never"]

for word in important_words:
    stop_words.discard(word)   # No error if word is absent

print("not" in stop_words)
```

**Text Data PreProcessing**

1. Converting into lower case

```python
text["tweet_text"] = text["tweet_text"].str.lower()
```

```python
text['tweet_text']
```

2. Removing all HTML tags

```python
# beautifulsoap is used to read html and xml tag and used to remove it from text
```

```python
from bs4 import BeautifulSoup

text["tweet_text"] = text["tweet_text"].apply(
    lambda x: BeautifulSoup(str(x), "html.parser").get_text()
)

print(text["tweet_text"].head())
```

3. Remove URL

```python
import re

text["tweet_text"] = text["tweet_text"].apply(
    lambda x: re.sub(r'https?://\S+|www\.\S+', '', str(x))
)

print(text["tweet_text"].head())
```

4. Removing only HashTag

```python
import re

text["tweet_text"] = text["tweet_text"].apply(
    lambda x: re.sub(r"#", "", str(x))
)

print(text["tweet_text"].head())
```

5. Removing Puncuation marks

```python
import string

# Remove all punctuation marks
text["tweet_text"] = text["tweet_text"].apply(
    lambda x: str(x).translate(str.maketrans('', '', string.punctuation))
)

print(text["tweet_text"].head())
```

6. Remove Special word

```python
import re

# Keep only alphabets, numbers and spaces
text["tweet_text"] = text["tweet_text"].apply(
    lambda x: re.sub(r'[^a-zA-Z0-9\s]', '', str(x))
)

print(text["tweet_text"].head())
```
