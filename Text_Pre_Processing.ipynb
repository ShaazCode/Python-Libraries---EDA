```python
text = "Hiiii!!!! I LOOOOOVE Python 😍😍 Visit www.python.org"
```

Ask
"What should we remove?"
Students answer.
Then do step by step.

```python
# step 01: converting from capital to small
text = text.lower()
print(text)
```

Ask
"What changed?"
Students
Capital became small.


```python
# step 02: removing URL
```

```python
import re
# Python doesn't know how to search patterns by default. So we import the re library.
# re = Regular Expression

text = re.sub(r'http\S+|www\S+', '', text)
# sub means subsitute - replace something with something
# syntax : # re.sub(pattern, replacement, text)
# pattern : what we are searching for ? website, phone number, email
# '' with what you are replace '' means nothing
# https : looks for word starting with https
# \S+ means 'keep taking every character until you reach a space

print(text)
```

Ask
"Why remove website?"
Expected answer
Website doesn't help understand sentiment.

Sentiment means the emotion, feeling, or opinion expressed in a piece of text.

```python
# step 03: Remove punctuation
```

```python
import string

text = text.translate(str.maketrans('', '', string.punctuation))
print(text)
# translate: it follows the rules which we created
# maketrnas : make - create trans - translate
# string.punctuation : This simply means: 'Give me all the punctuation symbols.' !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
```

```python
# step 04: Remove Numbers
```

```python
text1 = "shaaaz 1"
```

```python
text1 = re.sub(r'\d+', '', text1)
```

```python
# re.sub() → Replace or remove a pattern.
# r'' → Raw string (Python reads the pattern as it is).
# \d → Any digit (0–9).
# + → One or more digits.
# '' → Replace with nothing (remove it).
# text → The text where the replacement happens.
```

```python
text1
```

```python
# remove extra space
```

```python
text1 = " ".join(text1.split())

print(text1)
```

```python
# split() → Splits the sentence into words by removing extra spaces.
# " ".join() → Joins the words back together with one single space between them.
```

```python
from nltk.tokenize import word_tokenize
import nltk
# import the NLTK (Natural Language Toolkit) library.
# from nltk.tokenize → Go to the tokenize module inside the NLTK library.
# import word_tokenize → Import the function that splits a sentence into words (tokens).

nltk.download('punkt')
nltk.download('punkt_tab')

text = "I love learning NLP."

tokens = word_tokenize(text)

print(tokens)
```
